```python

import re
import time
import pickle
import random
import numpy as np
import nltk
import tensorflow as tf

from google.colab import files
from nltk.tokenize import sent_tokenize
from sklearn.model_selection import train_test_split
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# Settings
random.seed(42)
np.random.seed(42)
tf.random.set_seed(42)

MAX_WORDS = 6000
MAX_LEN = 15
EXAMPLES_PER_SENTENCE = 10
BATCH_SIZE = 128
EPOCHS = 10

print("Libraries imported successfully!")
```

```python

print("Choose your TXT dataset:")
uploaded = files.upload()

filename = next(iter(uploaded))

start = time.time()

with open(filename, "r", encoding="utf-8-sig") as f:
    text = f.read()

print("Dataset loaded:", filename)
print("Characters:", len(text))
print("File reading time:", round(time.time() - start, 2), "seconds")
```

```python

start = time.time()

# Use the TXT file already uploaded in Cell 2
text = text.strip()

sentences = [
    s.strip().lower()
    for s in text.splitlines()
    if len(s.split()) >= 3
]

if len(sentences) < 100:
    raise ValueError(
        f"Only {len(sentences)} sentences found. Check your TXT file."
    )

print("Total sentences:", len(sentences))
print("First sentence:", sentences[0])
print("Preprocessing time:", round(time.time() - start, 2), "seconds")
```

```python
from tensorflow.keras.layers import GRU

model = Sequential([
    Embedding(vocab_size, 64, mask_zero=True),
    GRU(128),
    Dropout(0.2),
    Dense(vocab_size, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.SparseTopKCategoricalAccuracy(
            k=5, name="top5_accuracy"
        )
    ]
)

model.summary()
```

```python

train_sentences, test_sentences = train_test_split(
    sentences,
    test_size=0.20,
    random_state=42
)

tokenizer = Tokenizer(
    num_words=MAX_WORDS,
    oov_token="<OOV>"
)

tokenizer.fit_on_texts(train_sentences)

vocab_size = min(MAX_WORDS, len(tokenizer.word_index) + 1)

print("Training sentences:", len(train_sentences))
print("Testing sentences:", len(test_sentences))
print("Vocabulary size:", vocab_size)
```

```python

def make_examples(sentence_list):
    X, y = [], []

    token_lists = tokenizer.texts_to_sequences(sentence_list)

    for tokens in token_lists:
        for i in range(1, len(tokens)):
            target = tokens[i]

            if target > 1:
                context = tokens[max(0, i - MAX_LEN):i]
                X.append(context)
                y.append(target)

    X = pad_sequences(
        X,
        maxlen=MAX_LEN,
        padding="pre",
        truncating="pre"
    )

    return X, np.array(y, dtype=np.int32)


start = time.time()

X_train, y_train = make_examples(train_sentences)
X_test, y_test = make_examples(test_sentences)

print("Training examples:", len(X_train))
print("Testing examples:", len(X_test))
print("Sequence preparation time:", round(time.time() - start, 2), "seconds")
```

```python

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=2,
    restore_best_weights=True
)

print("Training started...")
start = time.time()

history = model.fit(
    X_train,
    y_train,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.1,
    callbacks=[early_stop],
    verbose=1
)

training_time = time.time() - start

print("Training time:", round(training_time, 2), "seconds")
```

```python

model.save("next_word_model.keras")

with open("next_word_tokenizer.pkl", "wb") as f:
    pickle.dump(tokenizer, f)

print("Model saved: next_word_model.keras")
print("Tokenizer saved: next_word_tokenizer.pkl")
print("Program finished!")
```

```python

def predict_next_words(phrase, top_k=5):
    tokens = tokenizer.texts_to_sequences([phrase.lower()])[0]

    if not tokens or not any(t > 1 for t in tokens):
        print("No recognized words. Try another phrase.")
        return

    X = pad_sequences(
        [tokens[-MAX_LEN:]],
        maxlen=MAX_LEN,
        padding="pre",
        truncating="pre"
    )

    probabilities = model.predict(X, verbose=0)[0]
    ids = np.argsort(probabilities)[::-1]

    print("\nInput:", phrase)
    found = 0

    for idx in ids:
        word = tokenizer.index_word.get(int(idx))

        if word and word != "<OOV>" and idx != 0:
            print(
                f"{found + 1}. {word} "
                f"({probabilities[idx] * 100:.2f}%)"
            )
            found += 1

        if found >= top_k:
            break


# Warm-up
predict_next_words("the man was", top_k=3)

start = time.perf_counter()

for _ in range(10):
    model_input = pad_sequences(
        [tokenizer.texts_to_sequences(["the man was"])[0][-MAX_LEN:]],
        maxlen=MAX_LEN,
        padding="pre",
        truncating="pre"
    )
    model.predict(model_input, verbose=0)

average_latency = (time.perf_counter() - start) / 10

print(
    "\nAverage model prediction time:",
    round(average_latency * 1000, 2),
    "ms"
)

print("\n--- SAMPLE PREDICTIONS ---")

for phrase in [
    "i want to",
    "the man was",
    "she looked at",
    "there was a",
    "he said that"
]:
    predict_next_words(phrase)
```

```python
while True:
    phrase = input("Enter a phrase (or 'exit'): ")

    if phrase.lower().strip() == "exit":
        break

    predict_next_words(phrase)
```

```python

start = time.time()

results = model.evaluate(X_test, y_test, verbose=0)

evaluation_time = time.time() - start

print("\n--- EVALUATION RESULTS ---")
print(f"Test loss: {results[0]:.4f}")
print(f"Top-1 accuracy: {results[1] * 100:.2f}%")
print(f"Top-5 accuracy: {results[2] * 100:.2f}%")
print("Evaluation time:", round(evaluation_time, 2), "seconds")
```
